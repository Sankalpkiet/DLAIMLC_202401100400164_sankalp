```python
# Step 1: Given Data (1-D input)
X = [0, 1, 2, 3]     # 1-D matrix / vector
y = [0, 0, 1, 1]     # target output

# Initial weight, bias, learning rate
w = 0.0
b = 0.0
lr = 0.1
epochs = 10


# Step 3: Activation Function
def activation(z):
    return 1 if z >= 0 else 0


# Training
for epoch in range(epochs):
    print(f"\nEpoch {epoch+1}")
    for i in range(len(X)):

        # Step 2: Net input
        z = w * X[i] + b

        # Step 3: Prediction
        y_pred = activation(z)

        # Step 4: Error
        error = y[i] - y_pred

        # Step 5: Weight update
        w = w + lr * error * X[i]

        # Step 6: Bias update
        b = b + lr * error

        # Step 7: Print result
        print(f"X={X[i]}, Target={y[i]}, Pred={y_pred}, Error={error}")
        print(f"Updated w={w}, b={b}")


# Final Output
print("\nFinal Results:")
for i in range(len(X)):
    z = w * X[i] + b
    print(f"Input {X[i]} -> Output {activation(z)}")
```
